```python
# x=df_numeric.loc['wheelbase':'highwaympg']
x=df_numeric.iloc[:,2:15]
x
```

```python
# y=df_numric.loc['price']
y=df_numeric.iloc[:,15]
y
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=0)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
#scaling- to maintain balance b/w the digits of all column
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print("R² Score:", r2)
print("Model Accuracy (%):", r2 * 100)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
```

```python
x_train = pd.DataFrame(x_train, columns=x.columns)
x_test = pd.DataFrame(x_test, columns=x.columns)
```

```python
coefficients = pd.DataFrame({
    'Feature': x.columns,
    'Coefficient': model.coef_
})

coefficients
```

```python
# Positive coefficient → price increases
# Negative coefficient → price decreases
# Larger magnitude → stronger impact

# Coefficients represent the change in price for one-unit change in the feature,
# keeping other variables constant.
# sometimes removing outliers can reduce in number of rows which can impact model accuracy
# better to compare the results and then adopt the metods for outlier removal
```

```python
import matplotlib.pyplot as plt
y_pred = model.predict(x_test)
plt.scatter(y_test, y_pred)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted Car Prices')
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         color='red')
plt.show()

# Red line → perfect prediction
# Points close to line → good model
# Far away → large errors
# Actual vs predicted plot evaluates how well the model fits unseen data.
```
